# Task 1: Regex Fundamentals
Use regular expressions to extract:
1. All digits
2. Words starting with a capital letter
3. Sequences of exactly four digits
4. Words containing 3 to 6 alphabetic characters


```python
sample_text = (
    "Alice has 2 cats, Bob has 15 dogs, and Carol was born in 2004. "
    "David bought 7 books and used code 1234."
)

print("Sample Text:")
print(sample_text)
```

```python
# (a) Extract all digits
digits = re.findall(r"\d", sample_text)

print("All digits:")
print(digits)
```

```python
# (b) Extract all words that start with a capital letter
capital_words = re.findall(r"\b[A-Z][A-Za-z]*\b", sample_text)

print("Words starting with a capital letter:")
print(capital_words)
```

```python
# (c) Extract all sequences of exactly 4 digits
four_digit_sequences = re.findall(r"\b\d{4}\b", sample_text)

print("Exactly 4-digit sequences:")
print(four_digit_sequences)
```

```python
# (d) Extract words that are between 3 and 6 characters long
words_3_to_6 = re.findall(r"\b[A-Za-z]{3,6}\b", sample_text)

print("Words with 3 to 6 characters:")
print(words_3_to_6)
```

### Task 1 Explanation
- `\d` matches one digit.
- `[A-Z]` matches an uppercase English letter.
- `\d{4}` matches exactly four digits.
- `[A-Za-z]{3,6}` matches alphabetic words whose length is between 3 and 6 characters.
- `\b` is a word boundary and helps avoid partial matches.


# Task 2: Pattern-Based Extraction

The following paragraph contains emails, phone numbers, order IDs, dates, and percentages.  
We will extract the fields requested in the lab using `re.findall()`.


```python
paragraph = (
    "Customer Ali placed order ORD-1025 on 14/09/2026. "
    "His email is ali.khan25@example.com and phone number is 0301-1234567. "
    "Sara placed order ORD-2048 on 02/10/2026. "
    "Contact her at sara_22@shop.pk or 0333-7654321. "
    "The first order received a 45% discount while the second received 12.5%."
)

print(paragraph)
```

```python
# (a) Extract all email addresses
email_pattern = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b"
emails = re.findall(email_pattern, paragraph)

print("Email addresses:")
print(emails)
```

```python
# (b) Extract phone numbers in 03XX-XXXXXXX format
phone_pattern = r"\b03\d{2}-\d{7}\b"
phone_numbers = re.findall(phone_pattern, paragraph)

print("Phone numbers:")
print(phone_numbers)
```

```python
# (c) Extract dates in DD/MM/YYYY format
date_pattern = r"\b\d{2}/\d{2}/\d{4}\b"
dates = re.findall(date_pattern, paragraph)

print("Dates:")
print(dates)
```

```python
# (d) Extract percentage values such as 45% and 12.5%
percentage_pattern = r"\b\d+(?:\.\d+)?%" #(?: ... ) it grouos things together
percentages = re.findall(percentage_pattern, paragraph)

print("Percentages:")
print(percentages)
```

### Task 2 Explanation
- The email pattern checks the username, `@` sign, domain, and extension.
- The phone pattern follows the required Pakistani format `03XX-XXXXXXX`.
- The date pattern expects two digits, `/`, two digits, `/`, and four digits.
- `(?:\.\d+)?` makes the decimal part optional, so both `45%` and `12.5%` are matched.


# Task 3: Word Tokenization — NLTK vs spaCy

We use the same paragraph for both tokenizers. It contains contractions such as **don't**, **it's**, **can't**, and **he'll**, as well as an email address and a URL.


```python
text = (
    "I don't think it's difficult to learn tokenization. "
    "Dr. Ali can't attend today, but he'll join tomorrow. "
    "Visit https://example.com for more details. "
    "You can also email student@example.com for help."
)

print(text)
```

```python
# (a) Tokenize using NLTK word_tokenize()
nltk_tokens = word_tokenize(text)

print("NLTK Tokens:")
print(nltk_tokens)
print("\nTotal NLTK tokens:", len(nltk_tokens))
```

```python
# (b) Tokenize using a blank spaCy English pipeline
nlp_blank = spacy.blank("en")  #This creates a blank English spaCy pipeline/object
doc_blank = nlp_blank(text) #your text is passed into the spaCy pipeline.spaCy converts the text into a Doc object.A Doc is spaCy's processed version of the text.Inside it, the text is already split into tokens:

spacy_tokens = [token.text for token in doc_blank]

print("spaCy Tokens:")
print(spacy_tokens)
print("\nTotal spaCy tokens:", len(spacy_tokens))
```

```python
# (c) Comparison
print("Difference 1:")
print("spaCy normally keeps a URL such as https://example.com as one token,")
print("while NLTK may split parts of the URL depending on its tokenizer rules.")

print("\nDifference 2:")
print("spaCy recognizes email-like text such as student@example.com as one token,")
print("while NLTK may split it into multiple pieces.")

# Both tokenizers also apply their own rules to contractions and punctuation.
```

```python
# (d) Count the total tokens produced by each method
print("NLTK token count :", len(nltk_tokens))
print("spaCy token count :", len(spacy_tokens))

if len(nltk_tokens) == len(spacy_tokens):
    print("Both methods produced the same number of tokens.")
else:
    print("The two tokenizers produced different token counts.")
```

# Task 4: Token Attributes

We create a spaCy `Doc` containing:
- normal words,
- numbers,
- an email address,
- currency symbols.


```python
sentence = "Ali paid $50 and Sara paid €20 for 2 books. Email student@example.com for details."

doc = nlp_blank(sentence)

print(sentence)
```

```python
# (a), (b), and (c)
# Display token text with is_alpha, like_num, and like_email

print(f'{"Token":<25} {"is_alpha":<10} {"like_num":<10} {"like_email":<12}')
print("-" * 62)

for token in doc:
    print(
        f'{token.text:<25} '
        f'{str(token.is_alpha):<10} '
        f'{str(token.like_num):<10} '
        f'{str(token.like_email):<12}'
    )
```

```python
# (d) Identify only currency tokens
currency_tokens = [token.text for token in doc if token.is_currency]

print("Currency tokens:")
print(currency_tokens)
```

### Task 4 Observation
- `is_alpha` is `True` for tokens containing alphabetic letters only.
- `like_num` identifies number-like tokens.
- `like_email` identifies tokens that look like email addresses.
- `is_currency` identifies currency symbols such as `$` and `€`.


# Task 5: Sentence Segmentation — Handling Abbreviations

For this task, we use spaCy's trained English pipeline `en_core_web_sm` and check whether abbreviations such as **Dr.**, **Mr.**, and **e.g.** cause incorrect sentence breaks.


```python
# Load the trained spaCy model.
# If it is missing, install it automatically.

try:
    nlp = spacy.load("en_core_web_sm") #It tries to load spaCy’s English model.
except OSError:
    import subprocess
    import sys

    print("en_core_web_sm is not installed. Installing it now...")
    subprocess.check_call([
        sys.executable, "-m", "spacy", "download", "en_core_web_sm"
    ])
    nlp = spacy.load("en_core_web_sm")

print("Loaded pipeline:", nlp.pipe_names) #nlp.pipe_names tells you which NLP components are included in the loaded pipeline
```

```python
paragraph_task5 = (
    "Dr. Ahmed works at City Hospital. "
    "He sees patients every morning. "
    "Mr. Khan visited him today, e.g. for a routine check-up. "
    "Everything was normal."
)

doc_task5 = nlp(paragraph_task5)

sentences = list(doc_task5.sents)

print("Detected Sentences:")
for i, sent in enumerate(sentences, start=1):
    print(f"{i}. {sent.text}")

print("\nTotal number of sentences:", len(sentences))
```

```python
# (d) One-line observation
print(
    "Observation: The abbreviations Dr., Mr., and e.g. did not create an "
    "incorrect sentence break because spaCy handles common abbreviations using "
    "its trained language rules."
)
```
